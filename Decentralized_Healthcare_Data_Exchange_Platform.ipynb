```python
# ================================================================
# DECENTRALIZED HEALTHCARE DATA EXCHANGE PLATFORM
# ONE-CELL GOOGLE COLAB PROJECT VISUALIZATION
# Running this single cell automatically generates the image.
# ================================================================

import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch, Circle
import hashlib
import json
import os
from datetime import datetime

# ================================================================
# 1. SYNTHETIC HEALTHCARE DATA
# ================================================================

medical_record = {
    "patient_id": "P001",
    "record_type": "Lab Report",
    "date": "2026-08-26",
    "test": "Blood Test",
    "result": "Normal",
    "hospital": "Synthetic General Hospital",
    "data_type": "Dummy Educational Data"
}

# Create synthetic medical record
record_file = "medical_record_001.json"

with open(record_file, "w") as f:
    json.dump(medical_record, f, indent=4)

# ================================================================
# 2. SHA-256 HASH GENERATION
# ================================================================

with open(record_file, "rb") as f:
    file_data = f.read()

original_hash = hashlib.sha256(file_data).hexdigest()

record_id = "REC-" + original_hash[:8].upper()

# Simulated IPFS/off-chain reference
storage_reference = "ipfs://QmSyntheticHealthcareDemo001"

# ================================================================
# 3. SIMULATED BLOCKCHAIN DATA
# ================================================================

patient_address = "0xPATIENT001"
doctor_address = "0xDOCTOR001"
hospital_address = "0xHOSPITAL001"

blockchain_record = {
    "record_id": record_id,
    "patient": patient_address,
    "created_by": hospital_address,
    "record_type": medical_record["record_type"],
    "file_hash": original_hash,
    "storage_reference": storage_reference,
    "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
    "active": True
}

# ================================================================
# 4. CONSENT MANAGEMENT SIMULATION
# ================================================================

consent = {
    "patient": patient_address,
    "doctor": doctor_address,
    "purpose": "Treatment",
    "scope": "Read Lab Report",
    "status": "REVOKED"
}

audit_events = []

def add_event(action, actor, status):
    audit_events.append({
        "action": action,
        "actor": actor,
        "status": status,
        "time": datetime.now().strftime("%H:%M:%S")
    })

# Hospital adds medical record
add_event(
    "Medical Record Added",
    "Hospital",
    "SUCCESS"
)

# Doctor tries before consent
add_event(
    "Doctor Access Request",
    "Doctor",
    "DENIED"
)

# Patient grants access
consent["status"] = "GRANTED"

add_event(
    "Patient Grants Consent",
    "Patient",
    "SUCCESS"
)

# Doctor accesses record
add_event(
    "Authorized Record Access",
    "Doctor",
    "ALLOWED"
)

# Patient revokes access
consent["status"] = "REVOKED"

add_event(
    "Patient Revokes Consent",
    "Patient",
    "SUCCESS"
)

# Doctor tries again
add_event(
    "Doctor Access After Revocation",
    "Doctor",
    "DENIED"
)

# ================================================================
# 5. HASH VERIFICATION
# ================================================================

with open(record_file, "rb") as f:
    verification_hash = hashlib.sha256(f.read()).hexdigest()

hash_verified = original_hash == verification_hash

# ================================================================
# 6. CREATE PROFESSIONAL INFOGRAPHIC
# ================================================================

fig = plt.figure(figsize=(18, 12))
fig.patch.set_facecolor("#F4F7FB")

# Main drawing area
ax = fig.add_axes([0, 0, 1, 1])
ax.set_xlim(0, 100)
ax.set_ylim(0, 100)
ax.axis("off")

# ================================================================
# COLORS
# ================================================================

NAVY = "#102A43"
BLUE = "#1769AA"
LIGHT_BLUE = "#E8F3FC"
GREEN = "#16805C"
LIGHT_GREEN = "#E8F7F0"
RED = "#C0392B"
LIGHT_RED = "#FCECEA"
PURPLE = "#6C4AB6"
LIGHT_PURPLE = "#F0EBFA"
ORANGE = "#D97706"
LIGHT_ORANGE = "#FFF3E0"
WHITE = "#FFFFFF"
DARK = "#17202A"
GREY = "#52606D"
LIGHT_GREY = "#D9E2EC"

# ================================================================
# HELPER FUNCTIONS
# ================================================================

def rounded_box(x, y, w, h, title, body,
                facecolor=WHITE,
                edgecolor=LIGHT_GREY,
                title_color=NAVY,
                body_color=DARK,
                title_size=13,
                body_size=9.5):

    patch = FancyBboxPatch(
        (x, y), w, h,
        boxstyle="round,pad=0.5,rounding_size=2",
        linewidth=1.5,
        edgecolor=edgecolor,
        facecolor=facecolor
    )

    ax.add_patch(patch)

    ax.text(
        x + w/2,
        y + h - 3,
        title,
        ha="center",
        va="center",
        fontsize=title_size,
        fontweight="bold",
        color=title_color
    )

    ax.text(
        x + 2.5,
        y + h - 7,
        body,
        ha="left",
        va="top",
        fontsize=body_size,
        color=body_color,
        linespacing=1.55
    )


def arrow(x1, y1, x2, y2, color=BLUE, width=2):

    ax.add_patch(
        FancyArrowPatch(
            (x1, y1),
            (x2, y2),
            arrowstyle="-|>",
            mutation_scale=16,
            linewidth=width,
            color=color
        )
    )


def status_box(x, y, w, h, title, status, color):

    patch = FancyBboxPatch(
        (x, y), w, h,
        boxstyle="round,pad=0.5,rounding_size=2",
        linewidth=1.2,
        edgecolor=color,
        facecolor=WHITE
    )

    ax.add_patch(patch)

    ax.text(
        x + 2,
        y + h - 3,
        title,
        fontsize=9,
        fontweight="bold",
        color=GREY,
        va="top"
    )

    ax.text(
        x + w - 2,
        y + h - 3,
        status,
        fontsize=10,
        fontweight="bold",
        color=color,
        ha="right",
        va="top"
    )

# ================================================================
# HEADER
# ================================================================

ax.text(
    50, 96,
    "DECENTRALIZED HEALTHCARE DATA EXCHANGE PLATFORM",
    ha="center",
    va="center",
    fontsize=24,
    fontweight="bold",
    color=NAVY
)

ax.text(
    50, 92.8,
    "Blockchain-Based Consent Management • Secure Record Exchange • Hash Verification • Immutable Audit Trail",
    ha="center",
    va="center",
    fontsize=10.5,
    color=GREY
)

# Header line
ax.plot(
    [8, 92],
    [90.5, 90.5],
    linewidth=2,
    color=BLUE
)

# ================================================================
# ACTOR LAYER
# ================================================================

rounded_box(
    7, 70, 23, 15,
    "PATIENT",
    "• Owns healthcare records\n"
    "• Grants doctor access\n"
    "• Revokes consent\n"
    "• Controls data sharing",
    LIGHT_BLUE,
    BLUE
)

rounded_box(
    38.5, 70, 23, 15,
    "DOCTOR / PROVIDER",
    "• Requests access\n"
    "• Views authorized records\n"
    "• Cannot bypass consent\n"
    "• Uses permitted data",
    LIGHT_GREEN,
    GREEN
)

rounded_box(
    70, 70, 23, 15,
    "HOSPITAL",
    "• Registers record metadata\n"
    "• Generates record hash\n"
    "• Provides storage reference\n"
    "• Does not override consent",
    LIGHT_PURPLE,
    PURPLE
)

# Arrows between actors
arrow(30.2, 77.5, 38.0, 77.5, BLUE)
arrow(61.2, 77.5, 69.5, 77.5, PURPLE)

# ================================================================
# BLOCKCHAIN CORE
# ================================================================

rounded_box(
    7, 49, 86, 17,
    "BLOCKCHAIN / SMART CONTRACT LAYER",
    "ROLE REGISTRY        |        MEDICAL RECORD REGISTRY        |        CONSENT MANAGER        |        AUDIT EVENTS\n\n"
    "Patient / Doctor / Hospital roles     •     Record metadata     •     Access permissions     •     Consent grant/revoke\n"
    "File hash verification     •     Timestamp     •     Transaction/event history     •     Authorization checks",
    LIGHT_BLUE,
    BLUE,
    title_size=15,
    body_size=10
)

# ================================================================
# OFF-CHAIN STORAGE
# ================================================================

rounded_box(
    7, 31, 41, 14,
    "OFF-CHAIN / ENCRYPTED STORAGE",
    "Synthetic medical JSON / PDF\n"
    "Encrypted medical documents\n"
    "IPFS or local storage simulation\n"
    "Actual healthcare content remains off-chain",
    LIGHT_ORANGE,
    ORANGE,
    title_size=13,
    body_size=9.5
)

rounded_box(
    52, 31, 41, 14,
    "ON-CHAIN DATA",
    f"Record ID: {record_id}\n"
    f"Record Type: {medical_record['record_type']}\n"
    f"SHA-256: {original_hash[:20]}...\n"
    "Storage reference + timestamp + permissions",
    LIGHT_PURPLE,
    PURPLE,
    title_size=13,
    body_size=9
)

# Arrows storage ↔ blockchain
arrow(27.5, 45.2, 27.5, 48.5, ORANGE)
arrow(72.5, 45.2, 72.5, 48.5, PURPLE)

# ================================================================
# WORKFLOW
# ================================================================

ax.text(
    50, 27.5,
    "CONSENT-CENTRIC DATA FLOW",
    ha="center",
    va="center",
    fontsize=15,
    fontweight="bold",
    color=NAVY
)

workflow = [
    ("01", "Record\nCreated"),
    ("02", "Hash\nGenerated"),
    ("03", "Hash + Metadata\nStored On-Chain"),
    ("04", "Doctor\nRequests"),
    ("05", "Patient\nGrants Consent"),
    ("06", "Access\nAllowed"),
    ("07", "Patient\nRevokes"),
    ("08", "Access\nDenied")
]

start_x = 6
box_w = 10.2
gap = 1.3

for i, (num, text) in enumerate(workflow):

    x = start_x + i * (box_w + gap)

    circle = Circle(
        (x + box_w/2, 22.5),
        2.2,
        facecolor=BLUE,
        edgecolor=WHITE,
        linewidth=2
    )

    ax.add_patch(circle)

    ax.text(
        x + box_w/2,
        22.5,
        num,
        ha="center",
        va="center",
        fontsize=8,
        fontweight="bold",
        color=WHITE
    )

    ax.text(
        x + box_w/2,
        17.8,
        text,
        ha="center",
        va="center",
        fontsize=7.5,
        fontweight="bold",
        color=DARK
    )

    if i < len(workflow) - 1:
        arrow(
            x + box_w + 0.2,
            22.5,
            x + box_w + gap - 0.2,
            22.5,
            BLUE,
            1.4
        )

# ================================================================
# RESULT / PROOF SECTION
# ================================================================

ax.text(
    50, 13.5,
    "PROJECT EXECUTION RESULTS",
    ha="center",
    va="center",
    fontsize=14,
    fontweight="bold",
    color=NAVY
)

status_box(
    7, 8.5, 20, 3.5,
    "HASH INTEGRITY",
    "VERIFIED",
    GREEN
)

status_box(
    29, 8.5, 20, 3.5,
    "INITIAL ACCESS",
    "DENIED",
    RED
)

status_box(
    51, 8.5, 20, 3.5,
    "CONSENTED ACCESS",
    "ALLOWED",
    GREEN
)

status_box(
    73, 8.5, 20, 3.5,
    "POST-REVOCATION",
    "DENIED",
    RED
)

# ================================================================
# FOOTER
# ================================================================

ax.text(
    50, 4.8,
    "AUDIT EVENTS: 6    |    SYNTHETIC DATA ONLY    |    NO REAL PATIENT INFORMATION",
    ha="center",
    va="center",
    fontsize=9.5,
    fontweight="bold",
    color=GREY
)

ax.text(
    50, 2.5,
    "Educational Blockchain Prototype • Solidity / Hardhat Architecture • Off-Chain Medical Data • Patient-Controlled Consent",
    ha="center",
    va="center",
    fontsize=8.5,
    color=GREY
)

# ================================================================
# SAVE + DISPLAY IMAGE
# ================================================================

output_image = "Decentralized_Healthcare_Data_Exchange_Project.png"

plt.savefig(
    output_image,
    dpi=300,
    bbox_inches="tight",
    facecolor=fig.get_facecolor()
)

plt.show()

# ================================================================
# FINAL TERMINAL OUTPUT
# ================================================================

print("\n" + "=" * 75)
print("DECENTRALIZED HEALTHCARE DATA EXCHANGE PLATFORM")
print("=" * 75)
print("✓ Synthetic healthcare record created")
print("✓ SHA-256 hash generated")
print("✓ Blockchain record simulated")
print("✓ Patient / Doctor / Hospital roles simulated")
print("✓ Consent granted and revoked")
print("✓ Unauthorized access rejected")
print("✓ Hash integrity verified")
print("✓ Audit trail generated")
print("✓ Project architecture image generated")
print("=" * 75)
print(f"IMAGE FILE: {output_image}")
print("=" * 75)

# ================================================================
# AUTOMATIC COLAB DOWNLOAD
# ================================================================

try:
    from google.colab import files
    files.download(output_image)
except:
    print("Running outside Google Colab — image has been saved locally.")
```
